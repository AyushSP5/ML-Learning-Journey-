# Gradian decent class to create the proper function of gradient decent

```python
x,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1,noise=20,random_state=13)
```

```python
plt.scatter(x,y)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
lr.intercept_
```

```python
lr.coef_
```

```python
y_pred = lr.predict(x_test)
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```

```python
class mygradient:
    def __init__(self,learning_rate,epochs):
        self.b=-100
        self.m=100
        self.lr=learning_rate
        self.epoch=epochs

    def fit(self,x,y):
        for i in range(self.epoch):
            new_slope_b= -2* np.sum(y-self.m*x.ravel()-self.b)
            new_slope_m=-2* np.sum((y-self.m*x.ravel()-self.b)*x.ravel())
            
            self.b=self.b-(self.lr*new_slope_b)
            self.m=self.m-(self.lr*new_slope_m)
        print(self.b,self.m)

    def predict(self,x):
        return (self.m*x.ravel()+self.b)
```

```python
mg=mygradient(0.001,50)
```

```python
mg.fit(x,y)
```

```python
y_pred=mg.predict(x)
```

```python
from sklearn.metrics import r2_score
r2_score(y,y_pred)
```
