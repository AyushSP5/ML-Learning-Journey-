# Linear regration of Life Expectancy Data

```python
import pandas as pd
```

```python
df=pd.read_csv('Life Expectancy Data.csv')
```

```python
df.sample(6)
```

```python
df.columns = df.columns.str.strip()
```

```python
df=df.drop('Country',axis=1)
```

```python
df.shape
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
from sklearn.preprocessing import OrdinalEncoder

transform = ColumnTransformer(transformers=[
    ('tran1',
     SimpleImputer(),
    ['Adult Mortality','Alcohol','Hepatitis B','BMI','Polio', 'Total expenditure','Diphtheria', 'GDP', 'Population', 'thinness  1-19 years','thinness 5-9 years', 'Income composition of resources', 'Schooling']),
    ('tran2',OrdinalEncoder(categories=[['Developing','Developed']]),['Status'])
], remainder='passthrough')
```

```python
from sklearn.model_selection import train_test_split 
```

```python
x_train,x_test,y_train,y_test=train_test_split(df.drop('Life expectancy', axis=1),df['Life expectancy'],test_size=0.2,random_state=42)
```

```python
x_train_trf = transform.fit_transform(x_train)
```

```python
x_test_trf=transform.fit_transform(x_test)
```

```python
x_test_trf[0]
```

```python
transform.get_feature_names_out()
```

```python
columns = [
    'Adult Mortality',
    'Alcohol',
    'Hepatitis B',
    'BMI',
    'Polio',
    'Total expenditure',
    'Diphtheria',
    'GDP',
    'Population',
    'thinness  1-19 years',
    'thinness 5-9 years',
    'Income composition of resources',
    'Schooling',
    'Status',
    'Year',
    'infant deaths',
    'percentage expenditure',
    'Measles',
    'under-five deaths',
    'HIV/AIDS'
]
```

```python
x_train_trf = pd.DataFrame(
    transform.fit_transform(x_train),
    columns=columns,
    index=x_train.index
)
```

```python
x_test_trf = pd.DataFrame(
    transform.fit_transform(x_test),
    columns=columns,
    index=x_test.index
)
```

```python
x_train_trf.isnull().sum()
```

```python
lr=LinearRegression()
```

```python
y_test = y_test.fillna(y_test.mean())
```

```python
y_train = y_train.fillna(y_train.mean())
```

```python
lr.fit(x_train_trf,y_train)
```

```python
y_pred=lr.predict(x_test_trf)
```

# All the Matrics of the Model
