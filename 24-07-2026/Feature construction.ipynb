# Feature construction

```python
import pandas as pd 
import numpy as np 
```

```python
df = pd.read_csv('train.csv')[['Age','Pclass','SibSp','Parch','Survived']]
```

```python
df.head()
```

```python
df['Family members']=df['SibSp']+df['Parch']+1
```

```python
df
```

```python
def fun(x):
    if x == 1:
        return 0
    elif 1 < x < 4:
        return 1
    else:
        return 2
```

```python
df['Family_Category'] = df['Family members'].apply(fun)
```

```python
df.head()
```

```python
df.drop(columns=['SibSp','Parch','Family members'],inplace=True)
```
